```python
import os, glob

date_suffix = "_2026-10-28"

# 1. Rename Raw HTML files
for fp in glob.glob("raw/wikipedia_earthquake_list/*.html"):
    if date_suffix not in fp:
        os.rename(fp, fp.replace(".html", f"{date_suffix}.html"))

# 2. Rename Interim CSV files
for fp in glob.glob("interim/wikipedia_earthquake_list/*/*.csv"):
    if date_suffix not in fp:
        os.rename(fp, fp.replace(".csv", f"{date_suffix}.csv"))

print("All Wikipedia files successfully renamed!")
```
